- first we lower all the words and apply stopwords
- then we find the number of unique words and arrange them in the desending order of their frequency
- as the element which has come more number of time would have greater importance in the whole sentence

eg: 

1 - the boy is good 
2 - the girl is good
3 - the boy girl are good 

after applying stop word:

- 1 - boy good
- 2 - girl good
- 3 - boy girl good

we then arrange them in the desending order of their frequency:
- word                freq
- good                3
- boy                 2
- girl                2

- in binary bow we only use 1 and 0 but in case of bow we use the freq of word as the vector
