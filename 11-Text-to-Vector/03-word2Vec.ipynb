# CBOW Word2Vec (Continuous Bag of Words)

## Overview

CBOW (Continuous Bag of Words) is a neural network model used to learn word embeddings. It predicts a target word based on its surrounding context words. This approach helps capture semantic relationships between words by learning from the contexts in which words appear.

## How CBOW Works

### Step 1: Define the Window Size

We select a window size to determine which words will be used as context for training. The window size is always an **odd number** to ensure the target word is centered.

**Example:** With a window size of 5:
- 2 words before the target word
- The target word (center)
- 2 words after the target word
- Total context words: 4

### Step 2: Extract Context and Target Words

For each sentence in the corpus, we slide the window across and identify:
- **Input (Context words):** The words surrounding the target word
- **Output (Target word):** The word in the center of the window

**Example:**
Given the sentence: "the quick brown fox jumps"

With window size 5 centered on "brown":
- **Input context words:** [the, quick, fox, jumps]
- **Output target word:** brown

### Step 3: Convert Words to Vectors

Each context word is converted into its word vector (embedding). For the model's initial setup:
- We start with a vocabulary of all unique words in the corpus
- Each word is represented as a vector of fixed dimensions
- Initially, these vectors can be random or learned from the data

### Step 4: Process Through the Neural Network

The model processes the vectors as follows:

1. Take all context word vectors
2. Average them together to create a single combined context vector
3. Feed this averaged vector through a shallow neural network (ANN)
4. The network outputs predictions for what word should appear in that context position

### Step 5: Train the Model

The model learns by:
- Comparing its predictions with the actual target word
- Calculating the error (loss)
- Adjusting the weights in the neural network and updating word vectors to minimize this error
- Repeating this process across the entire corpus

## Output

After training, the model produces **word embeddings** where:
- Each word is represented as a dense vector of numbers
- Words that appear in similar contexts have similar vector representations
- These embeddings capture semantic and syntactic relationships between words
- The embeddings can be used for various NLP tasks (similarity, analogy, classification, etc.)

## Key Characteristics

- **Input:** Context words (surrounding words in the sentence)
- **Output:** Prediction of the target word
- **Window size:** Always odd-numbered for proper centering
- **Architecture:** Simple shallow neural network with one hidden layer
- **Result:** High-quality word embeddings that capture word meaning based on context

## Example Walkthrough

**Corpus sentence:** "machine learning is fascinating"

**Window size:** 3 (1 word before, target, 1 word after)

| Context Position | Input Words | Target Word |
|---|---|---|
| Around "learning" | [machine, is] | learning |
| Around "is" | [learning, fascinating] | is |

For each pair, the model learns to predict the target word from the context, and in doing so, learns meaningful word representations.
