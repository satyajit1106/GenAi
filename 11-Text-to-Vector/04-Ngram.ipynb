# N-grams: Definition and Examples

An **n-gram** is a contiguous sequence of **n** items (words, letters, or syllables) from a given sample of text or speech. N-grams are widely used in **natural language processing (NLP)** and **text mining** to analyze language patterns, build language models, and improve tasks like speech recognition, machine translation, and search engines.

---

## Types of N-grams

| Type      | Description                          | Example (from "The quick brown fox jumps over the lazy dog") |
|-----------|--------------------------------------|-------------------------------------------------------------|
| Unigram   | Single word                          | "The", "quick", "brown", "fox", "jumps", "over", "the", "lazy", "dog" |
| Bigram    | Two consecutive words                | "The quick", "quick brown", "brown fox", "fox jumps", "jumps over", ... |
| Trigram   | Three consecutive words              | "The quick brown", "quick brown fox", "brown fox jumps", ... |
| n-gram    | General term for any sequence of `n` items | "The quick brown fox", "quick brown fox jumps", ... |

---

## Why Use N-grams?

- **Language Modeling:** Predict the next word in a sentence (e.g., autocomplete).
- **Text Classification:** Identify spam, sentiment, or topics.
- **Machine Translation:** Improve translation accuracy by considering word context.
- **Speech Recognition:** Match spoken words to likely phrases.

---

## Example Use Case: Autocomplete

If a user types **"The quick brown"**, a language model trained on n-grams might suggest **"fox"** as the next word because the trigram **"The quick brown fox"** is common in English.

---

## Practical Example in Python

```python
from nltk import ngrams

sentence = "The quick brown fox jumps over the lazy dog"
words = sentence.split()

# Generate bigrams
bigrams = list(ngrams(words, 2))
print(bigrams)
# Output: [('The', 'quick'), ('quick', 'brown'), ('brown', 'fox'), ...]

# Generate trigrams
trigrams = list(ngrams(words, 3))
print(trigrams)
# Output: [('The', 'quick', 'brown'), ('quick', 'brown', 'fox'), ...]
